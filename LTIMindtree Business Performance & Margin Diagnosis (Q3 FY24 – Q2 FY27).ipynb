```python
import pandas as pd
```

```python
# 1. Load the dataset
df = pd.read_csv("LTIMindtree_Quarterly_Financial_Data.csv")
```

```python
print("rows and cols:", df.shape)
```

```python
print("cols:", list(df.columns))
```

```python
print("\nfirst few rows:")
print(df[["Fiscal_Quarter", "Sales_Cr", "Expenses_Cr", "OPM_Pct", "NPM_Pct", "Utilization_Pct"]].head())
```

```python
print("\nlast few rows:")
print(df[["Fiscal_Quarter", "Sales_Cr", "Expenses_Cr", "OPM_Pct", "NPM_Pct", "Utilization_Pct"]].tail())
```

```python
print("\nany nulls?")
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
print("\ninfo:")
df.info()
```

```python
print("\nquick summary:")
print(df[["Sales_Cr", "Expenses_Cr", "OPM_Pct", "NPM_Pct", "Utilization_Pct", "Headcount"]].describe().round(2))
```

```python
# clean up text fields and make a proper date column
df["Quarter_End"] = df["Quarter_End"].str.strip()
df["Fiscal_Quarter"] = df["Fiscal_Quarter"].str.strip()
df["Fiscal_Year"] = df["Fiscal_Year"].str.strip()
df["q_date"] = pd.to_datetime(df["Quarter_End"], format="%b %Y")
```

```python
# sort by time 
df = df.sort_values("q_date").reset_index(drop=True)
```

```python
# per-employee cost and operating profit (in lakhs)
df["cost_per_emp_lakhs"] = (df["Expenses_Cr"] * 100 / df["Headcount"]).round(2)
df["op_profit_per_emp_lakhs"] = (df["Operating_Profit_Cr"] * 100 / df["Headcount"]).round(2)
```

```python
df.to_csv("ltim_cleaned.csv", index=False)
print("\nsaved ltim_cleaned.csv", df.shape)
```
